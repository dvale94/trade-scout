```python
from trade_scout import config

from trade_scout.research.loader import load_trades
from trade_scout.research.metrics import (
    core_metrics, pretty_print_core,
    breakdown_expectancy, score_buckets, threshold_sweep
)
from trade_scout.research.mfe_mae import edge_diagnostics
from trade_scout.research.exit_simulator import compare_exit_policies
from trade_scout.research.plots import plot_distributions, equity_curve, scatter_mfe_mae
```

```python
trades = load_trades(config.LOG_DIR, market_type=config.MARKET_TYPE)

print("Rows:", len(trades))
display(trades.head(20))
```

```python
# What files are we expecting?
print("LOG_DIR:", config.LOG_DIR)
print("MARKET_TYPE:", config.MARKET_TYPE)

# Status coverage
if "exit_reason" in trades.columns:
    display(trades["exit_reason"].value_counts(dropna=False))

# How many are closed vs pending
closed = trades[trades["pnl_R"].notna()] if "pnl_R" in trades.columns else trades.iloc[0:0]
print("Closed trades:", len(closed))
```

```python
m = core_metrics(trades)
print(pretty_print_core(m))
```

```python
plot_distributions(trades)
```

```python
display(breakdown_expectancy(trades, by="tier").head(50))
display(breakdown_expectancy(trades, by="symbol").head(50))

if "anchor_regime" in trades.columns:
    display(breakdown_expectancy(trades, by="anchor_regime"))
```

```python
display(score_buckets(trades, buckets=10))

thresholds = [7.0, 7.2, 7.4, 7.6, 7.8, 8.0, 8.2, 8.4, 8.6, 8.8, 9.0]
display(threshold_sweep(trades, thresholds))
```

```python
diag = edge_diagnostics(trades)
diag
```

```python
display(compare_exit_policies(trades))
```

```python
equity_curve(trades)
scatter_mfe_mae(trades)
```
